## Creating Sample Data

Let's create a sample dataset to demonstrate Forge's capabilities:

```python
import pandas as pd
import numpy as np

# Set random seed for reproducibility
np.random.seed(42)

# Create sample data
n_samples = 1000

data = {
    'age': np.random.randint(18, 70, n_samples),
    'income': np.random.lognormal(10, 1, n_samples),
    'credit_score': np.random.randint(300, 850, n_samples),
    'category': np.random.choice(['A', 'B', 'C', 'D'], n_samples),
    'region': np.random.choice(['North', 'South', 'East', 'West'], n_samples),
}

X = pd.DataFrame(data)

# Create target variable (binary classification)
y = ((X['income'] > 30000) & (X['credit_score'] > 600)).astype(int)

print(f"Dataset shape: {X.shape}")
X.head()
```

## Automatic Feature Engineering with AutoFeatureTransformer

The `AutoFeatureTransformer` is the main entry point for automatic feature engineering:

```python
from forge import AutoFeatureTransformer

# Create transformer with automatic settings
transformer = AutoFeatureTransformer(
    max_features=50,  # Keep top 50 features
    verbose=1         # Show progress
)

# Fit and transform
X_engineered = transformer.fit_transform(X, y)

print(f"\nOriginal features: {X.shape[1]}")
print(f"Engineered features: {X_engineered.shape[1]}")
```

## Understanding Generated Features

Let's examine what features were created:

```python
# Get feature names
feature_names = transformer.get_feature_names_out()
print(f"Generated {len(feature_names)} features:")
for i, name in enumerate(feature_names[:20]):
    print(f"  {i+1}. {name}")
if len(feature_names) > 20:
    print(f"  ... and {len(feature_names) - 20} more")
```

## Feature Importance

Forge can compute feature importance to understand which features matter most:

```python
# Get feature importance
importance = transformer.get_feature_importance()

print("Top 10 Most Important Features:")
print(importance.head(10))
```

## Data Analysis

Forge also provides automatic data analysis:

```python
from forge import DataAnalyzer

analyzer = DataAnalyzer()
report = analyzer.analyze(X, y)

print("Column Types:")
for col, dtype in report.column_types.items():
    print(f"  {col}: {dtype}")

print("\nData Quality Issues:")
if report.quality_issues:
    for issue in report.quality_issues:
        print(f"  - {issue}")
else:
    print("  No issues found!")
```
